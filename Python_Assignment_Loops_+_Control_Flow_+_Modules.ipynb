D. for loop — Questions 11–15

```python
# Question 11: Print even numbers from 1 to 50
print("Even numbers from 1 to 50:")

for number in range(2, 51, 2):
    print(number, end=" ")

print()


# Question 12: Multiplication table from 1 to 12
n = int(input("Enter a number for its multiplication table: "))

for i in range(1, 13):
    print(n, "x", i, "=", n * i)


# Question 13: Sum of numbers from 1 to 100
total = 0

for number in range(1, 101):
    total += number

print("Sum from 1 to 100:", total)


# Question 14: Display lab equipment availability
equipment = [
    "Computer",
    "Keyboard",
    "Mouse",
    "Monitor",
    "Projector"
]

for item in equipment:
    print(item, "- Available")


# Question 15: Count scores above 50
count = 0

for i in range(1, 6):
    score = float(input(f"Enter score for student {i}: "))

    if score > 50:
        count += 1

print("Number of students scoring above 50:", count)
```

Explanation:
- range(2, 51, 2) generates even numbers from 2 to 50.
- range(1, 13) repeats the multiplication table 12 times.
- total += number adds each number to the running total.
- The equipment loop prints every item.
- The score loop counts only scores strictly greater than 50.

E. while loop — Questions 16–20

```python
# Question 16: Keep asking for the correct password
password = ""

while password != "lab2025":
    password = input("Enter password: ")

print("Correct password!")


# Question 17: Print numbers from 10 down to 1
number = 10

while number >= 1:
    print(number)
    number -= 1


# Question 18: Sum numbers until the user enters 0
total = 0

while True:
    number = float(input("Enter a number (0 to stop): "))

    if number == 0:
        break

    total += number

print("Total sum:", total)


# Question 19: Calculate factorial
n = int(input("Enter a non-negative integer: "))

if n < 0:
    print("Factorial is not defined for negative integers.")
else:
    factorial = 1
    number = 1

    while number <= n:
        factorial *= number
        number += 1

    print("Factorial of", n, "is", factorial)


# Question 20: Lab attendance
total_students = 0

while True:
    name = input("Enter student name (or type stop): ")

    if name.strip().lower() == "stop":
        break

    if name.strip() != "":
        total_students += 1

print("Total students in attendance:", total_students)
```

Explanation:
- Question 16 repeats until the password is lab2025.
- Question 17 decreases the number by one each time.
- Question 18 uses break to stop when the user enters 0.
- Question 19 multiplies numbers from 1 up to the entered integer. The factorial of 0 is 1.
- Question 20 counts names entered before the user types stop.

Part 2: Functions and Modules
You must create two files in the same folder. The following functions are original examples for lab activities and student records.

File 1 — file.py
Create a new Python file named exactly file.py and enter this code.

```python
# file.py
# Personal functions for lab activities and student records


# Function 1: Check whether a student passed
def check_student_pass(score):
    if score >= 50:
        return "Pass"
    else:
        return "Fail"


# Function 2: Generate a lab equipment ID
def generate_equipment_id(number):
    return f"FCP{number:03d}"


# Function 3: Calculate a student's average score
def calculate_average(scores):
    if len(scores) == 0:
        return 0

    return sum(scores) / len(scores)


# Function 4: Create a student record
def create_student_record(name, matric_no, department):
    record = {
        "name": name,
        "matric_no": matric_no,
        "department": department
    }

    return record
```

What the functions do:
Function	Purpose
check_student_pass(score)	Checks whether a score is at least 50.
generate_equipment_id(number)	Generates an equipment ID beginning with FCP.
calculate_average(scores)	Calculates the average of a list of scores.
create_student_record(...)	Creates a dictionary containing student details.
The return statement sends a result back to the code that calls the function.

File 2 — main.py
Create another file named main.py in the same folder. It imports file.py as a module and uses all four functions.

```python
# main.py
# Import the personal functions module
import file

print("=== PERSONAL FUNCTIONS AND MODULES ===")

# Function 1: Check student pass status
score = 75
result = file.check_student_pass(score)

print("Student score:", score)
print("Result:", result)


# Function 2: Generate a lab equipment ID
equipment_id = file.generate_equipment_id(7)

print("Generated equipment ID:", equipment_id)


# Function 3: Calculate average student score
scores = [65, 80, 75, 90, 70]
average = file.calculate_average(scores)

print("Student scores:", scores)
print("Average score:", average)


# Function 4: Create a student record
student = file.create_student_record(
    "Musa",
    "FUD/CYB/001",
    "Cyber Security"
)

print("Student record:", student)
```
