Part A: Variables and Data Types

Question 1: Create variables and print their types

```python
# Part A: Variables and Data Types

# Create variables using different data types
name = "Your Name"            # str (string)
age = 22                      # int (integer)
cgpa = 4.20                   # float (decimal number)
is_python_student = True      # bool (boolean)

# Print each variable and its data type
print("Name:", name)
print("Name type:", type(name))

print("Age:", age)
print("Age type:", type(age))

print("CGPA:", cgpa)
print("CGPA type:", type(cgpa))

print("Is Python student:", is_python_student)
print("Student status type:", type(is_python_student))
```

Part B: Lists

Question 2: Create a list called gadgets

```python
# Part B: Lists
# Question 2: Create a list of five electronic gadgets

gadgets = ["Laptop", "Phone", "Tablet", "Headphones", "Smartwatch"]

# Print the list
print("My gadgets:", gadgets)
```

Question 3: Perform list operations

```python
# Question 3a: Print the first and last gadgets
print("First gadget:", gadgets[0])
print("Last gadget:", gadgets[-1])

# Question 3b: Add two gadgets to the end of the list
gadgets.append("Camera")
gadgets.append("Printer")

print("After adding two gadgets:", gadgets)

# Question 3c: Remove one gadget from the middle
gadgets.remove("Tablet")

print("After removing Tablet:", gadgets)

# Question 3d: Print the final length of the list
print("Final length:", len(gadgets))
```

Part C: Tuples

Question 4: Create a tuple called coordinates

```python
# Part C: Tuples
# Question 4: Create a tuple containing three numbers

coordinates = (10, 20, 30)

# Print the tuple
print("Coordinates:", coordinates)
```

Question 5: Perform tuple operations

```python
# Question 5a: Try to change the second number to 99

try:
    coordinates[1] = 99
except TypeError as error:
    print("Error:", error)

# Explanation:
# Tuples are immutable, meaning their items cannot
# be changed after the tuple has been created.

# Question 5b: Access and print the last item
print("Last item:", coordinates[-1])
```

Explanation: A tuple is immutable. This means you cannot change, add, or remove its individual items after creating it. The try/except block catches the error so that the program continues running.

Part D: Dictionaries

Question 6: Create a dictionary called student

```python
# Part D: Dictionaries
# Question 6: Create a dictionary containing student details

student = {
    "name": "Your Name",
    "matric_no": "YOUR_MATRIC_NUMBER",
    "department": "Cyber Security",
    "age": 22
}

# Print the complete dictionary
print("Student details:", student)
```

Question 7: Perform dictionary operations

```python
# Question 7a: Print only the name and department
print("Student name:", student["name"])
print("Department:", student["department"])

# Question 7b: Add a new key-value pair for CGPA
student["cgpa"] = 4.50

# Question 7c: Update the student's age
student["age"] = 23

# Print the updated dictionary
print("Updated student details:", student)

# Question 7d: Print all keys separately
print("All keys:", list(student.keys()))

# Print all values separately
print("All values:", list(student.values()))
```

Explanation:
- A dictionary stores information as key-value pairs.
- student["name"] accesses a particular value.
- student["cgpa"] = 4.50 adds a new key-value pair.
- student["age"] = 23 updates an existing value.
- keys() returns the dictionary's keys.
- values() returns the dictionary's values.

Part E: Mixed Practice

Question 8: Create a list of dictionaries

```python
# Part E: Mixed Practice
# Question 8: Create a list containing three student dictionaries

class_list = [
    {"name": "Amina", "score": 85},
    {"name": "Musa", "score": 90},
    {"name": "Fatima", "score": 78}
]

# Print the complete class list
print("Class list:", class_list)

# Question 8a: Print the name of the second student
print("Second student's name:", class_list[1]["name"])

# Question 8b: Print the score of the last student
print("Last student's score:", class_list[-1]["score"])
```

Explanation:
- class_list contains three dictionaries.
- class_list[1] accesses the second student because Python indexing starts at 0.
- ["name"] retrieves that student's name.
- class_list[-1]["score"] retrieves the score of the last student.
